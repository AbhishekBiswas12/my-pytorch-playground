```python
# Zeros tensor
zero_tensor = torch.zeros(size=(3, 4))
```

Failed because python '*' operator just multiplies the values at each index of the tensors. torch.matmul does matrix multiplication (each element in the resultant matrix is the dot product of the corresponding row (from matrix1) and column (from matrix2))

## Reshape, transpose and permute

Run the following experiments:

1. Create a random tensor.
  - Reshape the tensor to a different shape and store in a different variable.
  - Change an element in the original tensor.
  - Check the same element in the reshaped variable. Expected -> The reshaped tensor changes too.
2. Create a random tensor.
  - Transpose the tensor to a different shape and store in a different variable.
  - Change an element in the original tensor.
  - Check the same element in the reshaped variable. Expected -> The transposed tensor doesn't change.
3. Create a random tensor.
  - Permute the tensor to a different shape and store in a different variable.
  - Change an element in the original tensor.
  - Check the same element in the reshaped variable. Expected -> The permuted tensor doesn't change.

- Permuted: rearrange the order of a tensor's dimensions (axes) without changing its underlying data

## Vectorization vs python loop

```python
m1 = torch.rand(100, 100)
m2 = torch.rand(100, 100)

m1_mini = torch.rand(3, 3)
m2_mini = torch.rand(3, 3)
```

Large matrix

```python
%%time
m1 * m2
```

```python
%%time
prod = []

for i in range(len(m1)):
  x = []
  for j in range(len(m1[0])):
    x.append(m1[i][j]*m2[i][j])
  prod.append(x)
```

Small matrix

```python
%%time
m1_mini * m2_mini
```

```python
%%time
prod = []

for i in range(len(m1_mini)):
  x = []
  for j in range(len(m1_mini[0])):
    x.append(m1_mini[i][j]*m2_mini[i][j])
  prod.append(x)
```

Observations:
- for large matrices, pytorch vector multiplication is considerably faster than python loops
- for smaller matrices, however, loops are pretty competitive with vectors
